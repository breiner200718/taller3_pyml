# Visión Artificial

## ¿Qué es la Visión Artificial?

La **visión artificial** es una tecnología que permite a las computadoras **analizar e interpretar imágenes y videos** para obtener información de ellos. Para esto, utiliza herramientas como la **inteligencia artificial, el procesamiento de imágenes y el aprendizaje automático**.

Su función principal es permitir que un sistema pueda reconocer diferentes elementos dentro de una imagen, como **objetos, personas, colores, formas, patrones y características**.

## ¿Dónde se utiliza?

La visión artificial tiene diferentes aplicaciones, por ejemplo:

- 🏥 **Medicina:** análisis de imágenes médicas.
- 🏭 **Industria:** revisión y control de productos.
- 🚗 **Automóviles:** detección de objetos y señales.
- 📱 **Aplicaciones:** reconocimiento de rostros y objetos.
- 🌱 **Agricultura:** identificación de plantas y enfermedades.

## Conclusión

En conclusión, la visión artificial permite que las computadoras **"vean" y analicen el contenido de imágenes o videos**, ayudando a automatizar tareas y tomar decisiones a partir de la información visual.

# Detección de Rostros con Python usando OpenCV

## ¿Qué es?

La **detección de rostros** es una técnica de visión artificial que permite identificar la presencia de caras humanas dentro de una imagen o video. En Python, una de las herramientas más utilizadas para realizar esta tarea es **OpenCV**.

OpenCV permite acceder a la cámara del computador, procesar las imágenes y detectar rostros utilizando diferentes algoritmos de visión artificial.

## ¿Cómo funciona?

El programa captura imágenes mediante la cámara y analiza cada fotograma en busca de características que correspondan a un rostro. Cuando encuentra uno, puede marcarlo utilizando un **rectángulo** alrededor de la cara.

## Herramientas utilizadas

- 🐍 **Python:** lenguaje utilizado para desarrollar el programa.
- 👁️ **OpenCV:** biblioteca utilizada para procesar imágenes y realizar la detección de rostros.
- 📷 **Cámara:** permite capturar las imágenes que serán analizadas.

## Objetivo

El objetivo es crear un programa capaz de **detectar rostros en tiempo real** utilizando la cámara del computador y las herramientas de visión artificial proporcionadas por OpenCV.

```python
# 1 instalacion de opencv library 

#pip install opencv-python

#2 importar el cv
import cv2

#cargar la imagen
img = 'faces.jpg'


#leer la imagen 
img = cv2.imread(img)
#5 mostrar la imagen
img.shape

print("version de opencv: ", cv2.__version__)
print("CascadeClassifier: ", hasattr(cv2.data, 'haarcascades'))
print(cv2.__file__)
```

```python
#6 convertir la imagen a escala de grises
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

gray_img.shape

```

```python
#cargqr el clasificador de caras

#7 cargar el clasificador de caras
cascade = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"

face_cascade = cv2.CascadeClassifier(cascade)

#8 verificar si se cargo correctamente el clasificador
if face_cascade.empty():
    print("No se pudo cargar el clasificador de caras")
else:
    print("Clasificador de caras cargado correctamente")
```

```python
#9 se realiza la deteccion de caras en la imagen
faces = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=5, minSize=(34, 34))
```

```python
#10 ford para reccorrer todas las caras y dinujar un rectangulo sobre la imagen original

for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)

#11 las iagenes se cargan en formato BGR, por lo que se debe convertir a RGB para mostrarla correctamente
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
#12 importar matplotlib para mostrar la imagen
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))
plt.imshow(img_rgb)
plt.axis('off')
```

```python
# 1. Acceder a la cámara de nuestro PC y leer un flujo de datos al tiempo real
video_capture = cv2.VideoCapture(0)

# función para detectar rostro con flujo de video y dibujar un recuadro delimitador
def detect_bounding_box(vid):
    gray_img = cv2.cvtColor(vid, cv2.COLOR_BGR2GRAY)
    face = face_cascade.detectMultiScale(
        gray_img, 
        1.1, 
        5, 
        minSize=(34, 34)
    )

    for (x, y, w, h) in face:
        cv2.rectangle(
            vid, 
            (x, y), 
            (x + w, y + h), 
            (0, 255, 0), 
            2
        )

    return face
```

```python

# bucle while para detectar caras en real time
while True:

    result, video_frame = video_capture.read()

    if result is False:
        break

    face = detect_bounding_box(video_frame)

    cv2.imshow("Mi cara detectada", video_frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break


# Estas líneas van FUERA del while
video_capture.release()
cv2.destroyAllWindows()
```
